```python
import pandas as pd
import seaborn as sns

df = pd.read_csv('static/data/losses.csv')
df.head()
```

```python

df.isnull().sum()
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
sns.boxplot(data=df)
plt.title('Cek Outliers')
plt.show()
```

```python
from scipy import stats
import numpy as np

# Menghitung Z-Score
z_scores = np.abs(stats.zscore(df))

# Menentukan threshold Z-Score (biasanya >3)
threshold = 3
df_no_outliers = df[(z_scores < threshold).all(axis=1)]

print(f"Data sebelum menghapus outliers: {len(df)}")
print(f"Data setelah menghapus outliers: {len(df_no_outliers)}")
```
